# Partie 2 : sous-échantillonner l'espace des k

```python
import numpy as np
from kspace_fig import load_kspace, make_frame, write_figure, write_static

K = load_kspace()
N1 = K.shape[1]
c1 = N1 // 2

def sous_ech(r):
    # on garde une colonne de phase sur r, centre compris
    garde = (np.arange(N1) - c1) % r == 0
    Kr = np.zeros_like(K)
    Kr[:, garde] = K[:, garde]
    return Kr
```

```python
frames = [make_frame(sous_ech(r), f"r = {r}") for r in range(1, 11)]
write_figure("../figures/fig-sous-echantillonnage.html", frames,
             slider_name="Facteur", title_k="log|K|", active=1)
write_static("../img/fig-sous-echantillonnage-static.png",
             [("r = 1", sous_ech(1)), ("r = 2", sous_ech(2))])
```
