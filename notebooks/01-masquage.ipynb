# Partie 1 : masquer le centre de l'espace des k

```python
import numpy as np
from kspace_fig import load_kspace, make_frame, write_figure, write_live_mask, write_static

K = load_kspace()
N0, N1 = K.shape
c0, c1 = N0 // 2, N1 // 2
E_tot = np.sum(np.abs(K) ** 2)

def masque(f, dy=0, dx=0):
    # rectangle de demi-côtés f % de la demi-matrice, centré en (c0 + dy, c1 + dx)
    h0, h1 = int(c0 * f / 100), int(c1 * f / 100)
    Km = K.copy()
    if h0 > 0 and h1 > 0:
        Km[max(0, c0 + dy - h0):c0 + dy + h0, max(0, c1 + dx - h1):c1 + dx + h1] = 0
    return Km
```

```python
frames = [make_frame(masque(f), f"{f} %") for f in range(0, 91)]
write_figure("../figures/fig-masquage.html", frames,
             slider_name="Taille du masque", title_k="log|K| masqué", active=20)
write_static("../img/fig-masquage-static.png",
             [("0 %", masque(0)), ("20 %", masque(20)), ("50 %", masque(50))])
```

```python
# masque déplaçable : la reconstruction est faite dans le navigateur
write_live_mask("../figures/fig-masquage-centre.html", K)
write_static("../img/fig-masquage-centre-static.png",
             [("centré", masque(15)), ("décalé", masque(15, 20, 10))])
```

```python
# tableau de la page 1
for f in [5, 10, 20, 30, 50, 90]:
    h0, h1 = int(c0 * f / 100), int(c1 * f / 100)
    surface = 4 * h0 * h1 / (N0 * N1)
    e = 1 - np.sum(np.abs(masque(f)) ** 2) / E_tot
    print(f"{f:>2} %   surface {100*surface:5.1f} %   énergie retirée {100*e:5.1f} %")
```
