# Espace des k : échelle linéaire ou logarithmique

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from kspace_fig import load_kspace, log_mag

K = load_kspace()
mag = np.abs(K)
print(f"max / min : {mag.max() / mag[mag > 0].min():.0f}")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(7, 5.2))
axs[0].imshow(mag, cmap="gray")
axs[0].set_title("|K| — échelle linéaire")
axs[1].imshow(log_mag(K), cmap="gray")
axs[1].set_title("log|K| — échelle logarithmique")
for ax in axs:
    ax.axis("off")
fig.tight_layout()

os.makedirs("../img", exist_ok=True)
fig.savefig("../img/kspace-lin-vs-log.png", dpi=180, bbox_inches="tight", facecolor="white")
```
